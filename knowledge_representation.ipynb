knowledge representation: - knowlewdge representation is a process of storing information about the real world in a form that the  computer can understand and reasonable
ex: real world knowledge

rules: tom is a cat, every cat is a mammal,every mammal is a animal.
tom is a cat,cat of x implies mammal of x

inference: tom is a animal,tom is a mammal

```python
# Knowledge Representation using Python

# --------------------------
# Facts
# --------------------------
facts = {"Cat(Tom)"}

# --------------------------
# Rules
# Key = Conclusion
# Value = Required Fact
# --------------------------
rules = {
    "Mammal(Tom)": "Cat(Tom)",
    "Animal(Tom)": "Mammal(Tom)"
}

print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nApplying Rules...")

# Keep applying rules until no new facts are added
changed = True

while changed:
    changed = False

    for conclusion, premise in rules.items():

        if premise in facts and conclusion not in facts:
            facts.add(conclusion)
            print(f"Derived: {conclusion}")
            changed = True

print("\nFinal Knowledge Base:")
for fact in facts:
    print(fact)
```
