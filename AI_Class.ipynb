TIC-TAC-TOE

```python
"""
Tic-Tac-Toe: Human vs AI
------------------------
You play X, the AI plays O.
The AI uses minimax with alpha-beta pruning, so on "hard" it never loses.

Run:
    python tic_tac_toe.py
"""

import math
import random

HUMAN = "X"
AI = "O"
EMPTY = " "

WIN_LINES = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),   # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),   # columns
    (0, 4, 8), (2, 4, 6),              # diagonals
]


def new_board():
    return [EMPTY] * 9


def print_board(board):
    rows = [board[i:i + 3] for i in range(0, 9, 3)]
    print()
    for r, row in enumerate(rows):
        print(" " + " | ".join(row))
        if r < 2:
            print("---+---+---")
    print()


def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != EMPTY and board[a] == board[b] == board[c]:
            return board[a]
    return None


def is_full(board):
    return EMPTY not in board


def empty_cells(board):
    return [i for i, v in enumerate(board) if v == EMPTY]


def game_over(board):
    return winner(board) is not None or is_full(board)


def minimax(board, depth, is_maximizing, alpha, beta):
    win = winner(board)
    if win == AI:
        return 10 - depth
    if win == HUMAN:
        return depth - 10
    if is_full(board):
        return 0

    if is_maximizing:
        best = -math.inf
        for i in empty_cells(board):
            board[i] = AI
            best = max(best, minimax(board, depth + 1, False, alpha, beta))
            board[i] = EMPTY
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = math.inf
        for i in empty_cells(board):
            board[i] = HUMAN
            best = min(best, minimax(board, depth + 1, True, alpha, beta))
            board[i] = EMPTY
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best


def best_move(board):
    best_score = -math.inf
    move = None
    for i in empty_cells(board):
        board[i] = AI
        score = minimax(board, 0, False, -math.inf, math.inf)
        board[i] = EMPTY
        if score > best_score:
            best_score = score
            move = i
    return move


def easy_move(board):
    # 60% random, 40% optimal -> beatable but not clueless
    if random.random() < 0.6:
        return random.choice(empty_cells(board))
    return best_move(board)


def ask_human_move(board):
    while True:
        raw = input("Your move (1-9): ").strip()
        if not raw.isdigit():
            print("Please enter a number from 1 to 9.")
            continue
        idx = int(raw) - 1
        if idx < 0 or idx > 8:
            print("Please enter a number from 1 to 9.")
            continue
        if board[idx] != EMPTY:
            print("That square is taken. Try again.")
            continue
        return idx


def choose_difficulty():
    print("Choose difficulty:")
    print("  1) Easy (beatable)")
    print("  2) Unbeatable")
    while True:
        choice = input("Enter 1 or 2: ").strip()
        if choice == "1":
            return "easy"
        if choice == "2":
            return "hard"
        print("Please enter 1 or 2.")


def play_round(difficulty):
    board = new_board()
    human_turn = random.choice([True, False])
    print("\nSquares are numbered 1-9, left to right, top to bottom.")
    print_board(["1", "2", "3", "4", "5", "6", "7", "8", "9"])
    print("You are X. AI is O.")
    print("You go first!" if human_turn else "AI goes first!")

    while not game_over(board):
        if human_turn:
            idx = ask_human_move(board)
            board[idx] = HUMAN
        else:
            print("AI is thinking...")
            idx = easy_move(board) if difficulty == "easy" else best_move(board)
            board[idx] = AI
        print_board(board)
        human_turn = not human_turn

    win = winner(board)
    if win == HUMAN:
        print("You win!")
        return "human"
    elif win == AI:
        print("AI wins!")
        return "ai"
    else:
        print("It's a draw.")
        return "draw"


def main():
    print("=== Tic-Tac-Toe: Human vs AI ===")
    difficulty = choose_difficulty()
    scores = {"human": 0, "ai": 0, "draw": 0}

    while True:
        result = play_round(difficulty)
        scores[result] += 1
        print(f"\nScore — You: {scores['human']}  AI: {scores['ai']}  Draws: {scores['draw']}")

        again = input("\nPlay again? (y/n): ").strip().lower()
        if again != "y":
            print("Thanks for playing!")
            break

        change = input("Change difficulty? (y/n): ").strip().lower()
        if change == "y":
            difficulty = choose_difficulty()


if __name__ == "__main__":
    main()
```

```python
from queue import PriorityQueue

# Graph representation
graph = {
    'A': [('B', 4), ('C', 3)],
    'B': [('D', 5), ('E', 12)],
    'C': [('E', 10)],
    'D': [('G', 7)],
    'E': [('G', 2)],
    'G': []
}

# Heuristic values (estimated distance to Goal G)
heuristic = {
    'A': 10,
    'B': 8,
    'C': 7,
    'D': 4,
    'E': 2,
    'G': 0
}

def astar(start, goal):
    open_list = PriorityQueue()
    open_list.put((0, start))

    g_cost = {start: 0}
    parent = {start: None}

    while not open_list.empty():
        current = open_list.get()[1]

        if current == goal:
            path = []
            while current:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for neighbor, cost in graph[current]:
            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                f_cost = new_g + heuristic[neighbor]

                open_list.put((f_cost, neighbor))
                parent[neighbor] = current

    return None

# Run A*
path = astar('A', 'G')

print("Shortest Path:", " -> ".join(path))
```

```python
# Graph
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2)],
    'C': [('D', 1)],
    'D': []
}

# Heuristic values
h = {
    'A': 3,
    'B': 2,
    'C': 1,
    'D': 0
}

start = 'A'
goal = 'D'

open_list = [(start, 0)]  # (node, g_cost)
visited = []

while open_list:

    # Select node with minimum f = g + h
    current, g = min(open_list, key=lambda x: x[1] + h[x[0]])
    open_list.remove((current, g))

    print("Visiting:", current)

    if current == goal:
        print("Goal Reached!")
        break

    visited.append(current)

    for neighbor, cost in graph[current]:
        if neighbor not in visited:
            open_list.append((neighbor, g + cost))
```

```python
rooms = ["Room1", "Room2", "Room3", "Room4", "Room5"]
subjects = ["AI", "DBMS", "OS"]

conflicts = {
    "Room1": ["Room2", "Room3"],
    "Room2": ["Room1", "Room3", "Room4"],
    "Room3": ["Room1", "Room2", "Room5"],
    "Room4": ["Room2", "Room5"],
    "Room5": ["Room3", "Room4"],
}

def is_valid(room, subject, assignment):
    for n in conflicts.get(room, []):
        if assignment.get(n) == subject:
            return False
    return True

def backtrack(assignment):
    if len(assignment) == len(rooms):
        return assignment
    room = [r for r in rooms if r not in assignment][0]
    for subject in subjects:
        if is_valid(room, subject, assignment):
            assignment[room] = subject
            result = backtrack(assignment)
            if result:
                return result
            del assignment[room]
    return None

allocation = backtrack({})
for room in rooms:
    print(room, "->", allocation[room])
```

```python
N = 8

board = [[0 for _ in range(N)] for _ in range(N)]

def is_safe(board, row, col):

    # Check left side
    for i in range(col):
        if board[row][i] == 1:
            return False

    # Upper diagonal
    i, j = row, col
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Lower diagonal
    i, j = row, col
    while i < N and j >= 0:
        if board[i][j] == 1:
            return False
        i += 1
        j -= 1

    return True

def solve(board, col):

    if col >= N:
        return True

    for row in range(N):

        if is_safe(board, row, col):

            board[row][col] = 1

            if solve(board, col + 1):
                return True

            board[row][col] = 0

    return False

if solve(board, 0):

    print("Solution Found:")
    for row in board:
        print(row)

else:
    print("No Solution Exists")
```

```python
# Variables (Students)
students = ["A", "B", "C"]

# Domain (Available Seats)
seats = [1, 2, 3]

# Dictionary to store assignments
assignment = {}

# Function to check whether assigning a seat is valid
def is_valid(student, seat):
    # Constraint 1: A cannot sit in Seat 2
    if student == "A" and seat == 2:
        return False

    # Constraint 2: No two students can have the same seat
    if seat in assignment.values():
        return False

    return True

# Backtracking algorithm
def backtrack(index):
    # If all students are assigned
    if index == len(students):
        return True

    student = students[index]

    # Try every seat
    for seat in seats:
        if is_valid(student, seat):
            assignment[student] = seat

            if backtrack(index + 1):
                return True

            # Backtrack
            del assignment[student]

    return False

# Solve the CSP
if backtrack(0):
    print("Solution Found:")
    for student, seat in assignment.items():
        print(f"{student} --> Seat {seat}")
else:
    print("No solution exists.")
```

Backward chaining


```python
# -------------------------------
# Backward Chaining Example
# -------------------------------

# Facts stored in knowledge base
facts = {
    "Human(Socrates)"
}

# Rules stored as:
# conclusion : premise
rules = {
    "Mortal(Socrates)": "Human(Socrates)"
}

# Recursive backward chaining function
def backward_chain(goal):

    print(f"\nTrying to prove: {goal}")

    # Step 1: Check whether goal is already a fact
    if goal in facts:
        print(f"{goal} is a FACT.")
        return True

    # Step 2: Check if any rule can prove it
    if goal in rules:

        premise = rules[goal]

        print(f"Need to prove: {premise}")

        if backward_chain(premise):
            print(f"Rule applied:")
            print(f"{premise} --> {goal}")
            return True

    print(f"Cannot prove {goal}")
    return False


# Goal
goal = "Mortal(Socrates)"

print("Backward Chaining")

if backward_chain(goal):
    print("\nGoal PROVED")
else:
    print("\nGoal NOT PROVED")
```

```python
# Backward Chaining Example

# Known facts
facts = {"Study"}

# Rule: Pass requires Study
rules = {
    "Pass": "Study"
}

# Goal
goal = "Pass"

print("Goal:", goal)

if goal in rules:
    required_fact = rules[goal]
    print("Need to prove:", required_fact)

    if required_fact in facts:
        print("Fact found:", required_fact)
        print("Goal Proved")
    else:
        print("Goal Cannot Be Proved")
else:
    print("No rule available")
```

rules: tom is a cat, every cat is a mammal,every mammal is a animal.

inference: tom is a animal,tom is a mammal

```python
# Knowledge Representation using Python

# --------------------------
# Facts
# --------------------------
facts = {"Cat(Tom)"}

# --------------------------
# Rules
# Key = Conclusion
# Value = Required Fact
# --------------------------
rules = {
    "Mammal(Tom)": "Cat(Tom)",
    "Animal(Tom)": "Mammal(Tom)"
}

print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nApplying Rules...")

# Keep applying rules until no new facts are added
changed = True

while changed:
    changed = False

    for conclusion, premise in rules.items():

        if premise in facts and conclusion not in facts:
            facts.add(conclusion)
            print(f"Derived: {conclusion}")
            changed = True

print("\nFinal Knowledge Base:")
for fact in facts:
    print(fact)
```
