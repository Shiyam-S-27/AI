```python
# Medical Diagnosis

diseases = {
    "flu": ["fever", "cough", "body_pain"],
    "cold": ["cough", "runny_nose", "sneezing"],
    "malaria": ["fever", "chills", "sweating"],
    "covid": ["fever", "cough", "loss_of_taste"]
}

def diagnose(symptoms):
    for disease, required_symptoms in diseases.items():
        if all(symptom in symptoms for symptom in required_symptoms):
            print("Diagnosis:", disease)

# Query
symptoms = ["fever", "cough", "body_pain"]

print("Symptoms:", symptoms)
diagnose(symptoms)
```
